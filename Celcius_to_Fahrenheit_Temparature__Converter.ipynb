#Temperature Conversion and Threshold Checker
This notebook converts temperatures from Celsius (°C) to Fahrenheit (°F), checks a temperature against a specified threshold, and demonstrates the use of functions, loops, and user input.

```python
# Function to check body temperature and convert Celsius to Fahrenheit
def temperature_checker():

    # Nested function for temperature conversion
    def celsius_to_fahrenheit(celsius):
        return (celsius * 9/5) + 32

    # Normal body temperature reference for threshhold comparison
    normal_temperature = 37.0

    # Continue running on loop until the user chooses to exit
    while True:
        while True:
            try:
                #The try blocks validates user input and stops the function
                #from breaking when a non numeric input is entered
                #Get temperature from the user. Escape character improves code
                #readability
                celsius = float(input("\nEnter your body temperature in °C: "))
            except ValueError:
                print("Invalid input. Please enter a valid number.")
            else:
                break

        # Calls user defined function to Convert Celsius to Fahrenheit
        fahrenheit = celsius_to_fahrenheit(celsius)

        # Display the temperatures
        print(f"\nTemperature in Celcius: {celsius:.2f}°C")
        print(f"Temperature in Fahrenheits: {fahrenheit:.2f}°F")

        # Check for possible fever. Compares temparature to a defined threshhold
        if celsius > normal_temperature:
            print("You may have a fever.")
        elif celsius < normal_temperature:
            print("Temperature is below the normal reference temperature.")
        else:
            print("Temperature is at the normal reference temperature.")

        # Ask whether the user wants to continue
        #New variable choice stores user input, strip() removes leading or
        #trailing spaces while lower converts input to lower case
        #This loop ensures a valid input is provided for function to proceed
        #This block also catches invalid input to prevent function from breaking
        while True:
            choice = input(
                "\nWould you like to convert another temperature? (yes/no): "
            ).strip().lower()

            if choice == "yes":
                break
            elif choice == "no":
                print("\nThank you for using the temperature checker. Goodbye!")
                return
            else:
                print("Invalid choice. Please enter only 'yes' or 'no'.")

# Call the main function
temperature_checker()
```
